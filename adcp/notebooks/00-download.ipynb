# 00 — Download
Download raw ADCP and co-located CTD data from OOINet and assemble ship bottle data.

```python
import sys, yaml
sys.path.insert(0, '..')
from adcp import download
```

```python
# Load instrument config
config = yaml.safe_load(open('../config/GI01SUMO-RII11-02-ADCPSN010.yml'))
refdes = config['refdes']
print(refdes)
```

## ADCP streams
Downloads telemetered, recovered_host, and recovered_inst streams plus the
telemetered metadata stream.  Serial numbers are assigned per deployment.

```python
adcp_paths = download.download_adcp(config, output_dir='../data/')
adcp_paths
```

## Co-located CTD
Downloads and merges the three CTD streams for the mooring.

```python
ctd_path = download.download_ctd(config, output_dir='../data/')
print(ctd_path)
```

```python
ship_dir = '/path/to/cruise_data/irminger/'   # ← update this path
bottle_path = download.download_bottle_data(ship_dir, output_dir='../data/')
print(bottle_path)
```
